```python
# [Cell 1] - Import Library & Load Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
%matplotlib inline

df = pd.read_csv('/content/Delivery_Logistics.csv')

# [Cell 2] - Multivariat: Performa Mitra Ekspedisi vs Target Status
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='delivery_partner', hue='delivery_status', palette='viridis')
plt.title('Hubungan Mitra Ekspedisi dengan Status Pengiriman', fontsize=14)
plt.xlabel('Mitra Ekspedisi')
plt.ylabel('Jumlah Paket')
plt.xticks(rotation=45)
plt.legend(title='Status')
plt.tight_layout()
plt.show()

# [Cell 3] - Multivariat: Kondisi Cuaca vs Status Pengiriman
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='weather_condition', hue='delivery_status', palette='magma')
plt.title('Pengaruh Kondisi Cuaca terhadap Status Keterlambatan', fontsize=14)
plt.xlabel('Kondisi Cuaca')
plt.ylabel('Jumlah Paket')
plt.legend(title='Status')
plt.show()

# [Cell 4] - Multivariat: Korelasi Matriks Antar Fitur Numerik
plt.figure(figsize=(8, 6))
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
sns.heatmap(df[num_cols].corr(), annot=True, cmap='Blues', fmt='.2f')
plt.title('Matriks Korelasi Antar Fitur Numerik', fontsize=14)
plt.show()
```
